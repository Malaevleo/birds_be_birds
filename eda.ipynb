```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel("lq_only.xlsx")
data.head()
```

```python
data.describe()
```

```python
data = data.sort_values(by=["LQ"])
data.head(20)
```

```python
low_lq = data[data["LQ"] < 0.99]
mid_lq = data[(data["LQ"] >= 0.99) & (data["LQ"] < 2)]
high_lq = data[data["LQ"] >= 2]
print(f"Low LQ species: {len(low_lq)}")
print(f"Mid LQ species: {len(mid_lq)}")
print(f"High LQ species: {len(high_lq)}")
```

```python
species_counts = [len(low_lq), len(mid_lq), len(high_lq)]
sns.barplot(x=["Low LQ", "Mid LQ", "High LQ"], y=species_counts)
plt.show()
```
